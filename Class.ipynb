```python
df.groupby('parental level of education')['math score'].mean()
```

```python
df['average score'] = df[['math score', 'reading score', 'writing score']].mean(axis=1)
```

```python
df['passed'] = df['average score'] >= 60
```

```python
df['passed'].value_counts()
```
